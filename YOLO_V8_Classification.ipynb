```python
import os
import yaml

def generate_data_yaml(output_dir, yaml_path="data.yaml"):
    """
    Generates a data.yaml file for YOLO classification.

    Args:
        output_dir (str): Path to the dataset output directory containing 'train' and 'val'.
        yaml_path (str): Path to save the generated data.yaml file.

    Returns:
        None
    """
    # Paths for train and val folders
    train_path = os.path.join(output_dir, "train")
    val_path = os.path.join(output_dir, "val")

    # Get class names from the 'train' folder
    class_names = sorted(os.listdir(train_path))

    # Define the YAML structure
    data = {
        'train': train_path,
        'val': val_path,
        'nc': len(class_names),
        'names': class_names
    }

    # Save to a YAML file
    with open(yaml_path, 'w') as file:
        yaml.dump(data, file, default_flow_style=False)

    print(f"data.yaml generated successfully at {yaml_path}")

# Example usage
output_path = "/content/dataset/data_split"  # Path to your train/val split directory
yaml_output_path = "/content/dataset/data_splitdata.yaml"  # Path to save data.yaml
generate_data_yaml(output_path, yaml_output_path)
```
